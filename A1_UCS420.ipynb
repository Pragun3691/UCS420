Assignment 5.3: WAP to add all prime numbers from 1 to n and n is user input

```python
num1 = int(input("Enter number :"))
print("Sum of all prime numbers from 1 to number is : ")
flag = 0
sum = 0
for j in range(1,num1+1):
    flag = 0
    if j < 2:
      continue
    flag = 1
    for i in range(2,int(j/2)+1):
      if j%i==0:
        flag = 0
        break
    if flag == 1:
      sum+=j

print(sum)
```

Assignment 6.1 : WAP using function that add all odd numbers from 1 to n, n is user input

```python
n = int(input("Enter number :"))
def oddsum(n):
  sum = 0
  for i in range(1,n+1):
    if(i%2!=0):
      sum+=i
  return sum

print("Sum of odd numbers from 1 to", n, "is : ", oddsum(n))
```

Assignment 6.2 : WAP using function that add all prime numbers from 1 to n, n given by the user

```python
num1 = int(input("Enter number :"))
def primesum(num1):
  flag = 0
  sum = 0
  for j in range(1,num1+1):
     flag = 0
     if j < 2:
        continue
     flag = 1
     for i in range(2,int(j/2)+1):
        if j%i==0:
         flag = 0
         break
     if flag == 1:
        sum+=j
  return sum

print("Sum of prime numbers from 1 to", num1, "is : ", primesum(num1))
```
