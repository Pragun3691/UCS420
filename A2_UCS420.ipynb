3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating
your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed
produces a different list, so this cannot be copied from a classmate).

```python
import random
random.seed(int(roll_no))
```

```python
# i. Generate a list of 100 random numbers between 100 and 900 (inclusive).
numbers = [random.randint(100, 900) for _ in range(100)]
print(numbers)
```

```python
# ii. Count and print all odd numbers in the list.
odd_numbers = [x for x in numbers if x % 2 != 0]
print("Odd numbers count:", len(odd_numbers))
print("Odd numbers:", odd_numbers)
```

```python
# iii. Count and print all even numbers in the list.
even_numbers = [x for x in numbers if x % 2 == 0]
print("Even numbers count:", len(even_numbers))
print("Even numbers:", even_numbers)
```

```python
# iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers
# found using list comprehension.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n//2) + 1):
        if n % i == 0:
            return False
    return True
prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers count:", len(prime_numbers))
print("Prime numbers:", prime_numbers)
```

```python
# v. Print the number that occurs most frequently in your list, and how many times it occurs
most_frequent = max(numbers, key=numbers.count)
print("Most frequent number:", most_frequent)
print("It occurs:", numbers.count(most_frequent), "times")
```

4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
roll_no = "1024160094"
digits = [int(d) for d in roll_no[:8]]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print("Set A:", A)
print("Set B:", B)
```

```python
# vi. Find the union of A and B — all unique values across both sets.
union = A.union(B)
print("Union of A and B:", union)
```

```python
# vii. Find the intersection of A and B — values common to both.
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)
```

```python
# viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
# differs from symmetric_difference().
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```python
# ix. Find the symmetric difference of A and B.
print("Symmetric difference:", A.symmetric_difference(B))
```

```python
# x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
# xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
# in one line why discard() is safer than remove() when you're not sure the value exists.
X = int(input("Enter a value to remove from A: "))
A.discard(X)
print(A)
```
